```python
import pandas as pd
import numpy as np
```

## Part 1

```python
participants = pd.DataFrame({
    "id": [101, 102, 103, 104, 105, 106],
    "age": [24, 31, 27, 19, 42, 35],
    "group": ["A", "B", "A", "B", "A", "B"]
})
```

```python
scores = pd.DataFrame({
    "id": [101, 102, 103, 105, 106, 107],
    "score": [81, 94, 88, 91, 85, 79]
})
```

```python
display(participants)
display(scores)
```

## Part 2 Inner Join

```python
inner = participants.merge(
    scores,
    on="id",
    how="inner"
)
```

```python
display(inner)
inner.shape
```

Questions

Which IDs were removed?

104 and 107

Why?

Because an inner join joins the dataframe on the intersection of the specified column of both dataframes.

Yes

## Part 3

```python
left = participants.merge(
    scores,
    on="id",
    how="left"
)

left
```

Questions

Which table defines the resulting population?

What happened to ID 104?

Why is its score missing?

Because it is not defined in the scores dataframe.

## Part 4 — Outer join

```python
outer = participants.merge(
    scores,
    on="id",
    how="outer"
)
outer
```

It will find the union of the two datasets

```python
right = participants.merge(
    scores,
    on="id",
    how="right"
)

right
```

## Part 5 — Diagnose joins with indicator=True

```python
diagnostic = participants.merge(
    scores,
    on="id",
    how="outer",
    indicator=True
)

diagnostic
```
