```python
def add_value(values):
    values.append(100)

data = [1, 2, 3]
add_value(data)

print(data)
```

```python
def add_value(values):
    result = values.copy()
    result.append(100)
    return result

data = [1, 2, 3]
new_data = add_value(data)

print(data)
print(new_data)
```

I would prefer the non-mutating version because it preserves the original data that was fed into it.

Write the equivalent traditional loop and compare readability.

```python
even_squares = []
for x in range(21):
    if x % 2 == 0:
        even_squares.append(x ** 2)

even_squares
```

The traditional loop is likely more readiable because the steps are in chronological order where as with the list comprehension version, the result of the if statement is read first, then the loop condition, and then the if statement.
