## Part 7

````python
def simulate_scores(
    n,
    mean,
    sd,
    seed
):
    rng = np.random.default_rng(seed)

    return rng.normal(
        mean,
        sd,
        n
    )
````

````python
simulate_scores(
    100,
    80,
    10,
    556
)
````

````python
simulate_scores(
    n=100,
    mean=80,
    sd=10,
    seed=556
)
````

Which is easier to interpret?

The second funciton call is easier to interpret because the parameters are specified making it easy to understand what each value means

## Part 17

````python
x = "global"
````

````python
def outer():
    x = "enclosing"

    def inner():
        x = "local"
        return x

    return inner()
````

````python
outer()
````

Remove the local x from inner() and repeat.

````python
def outer():
    x = "enclosing"

    def inner():
        return x

    return inner()

outer()
````

Then remove the enclosing x.

````python
def outer():

    def inner():
        return x

    return inner()

outer()
````

Complete:

L: local 

E: enclosing

G: global

B: Built-in

Explain the lookup order.

The lookup order follows a heirarchy. It first looks locally, then enclosing, then globally, and the built-in.

## Part 18

````python
def analyze(data):
    clean = data.dropna(
        subset=["age", "score"]
    )

    clean = clean.loc[
        clean["age"] >= 18
    ]

    mean = clean["score"].mean()

    return mean
````

````python
def remove_missing(data):
    clean = data.dropna(
        subset=['age', 'score']
    )
    return clean
````

````python
def select_adults(data):
    clean = data.loc[data['score'] >= 18]
    return clean
````

````python
def mean_score(data):
    mean = data['score'].mean()
    return mean
````
