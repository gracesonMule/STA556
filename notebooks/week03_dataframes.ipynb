```python
def select_high_scores(
    data: pd.DataFrame,
    minimum_score: float
) -> pd.DataFrame:
    return data.loc[data['score']>=minimum_score]
```

```python
select_high_scores(participants, 85)
```

```python
select_high_scores(participants, 90)
```

Extension

Modify the function so the caller can also specify which columns to return.

```python
def select_high_scores(
    data: pd.DataFrame,
    minimum_score: float,
    specific_columns: list
) -> pd.DataFrame:
    if specific_columns:
        return data.loc[data['score']>=minimum_score, specific_columns]
    
    return data.loc[data['score']>=minimum_score]
```

```python
select_high_scores(participants, 90, 'age')
```
