```python
students_enrollment = students_df.merge(enrollment_df, on='student_id', how='left', validate='one_to_many', indicator=True)
students_enrollment.loc[~(students_enrollment['_merge']=='both'), ['_merge']]
```

After merging students with enrollment, only one entry was found to not exist in both.

```python
students_enrollment_courses = courses_df.merge(students_enrollment, on='course_id', how='left', validate='one_to_many')
students_enrollment_courses.head(5)
```

When merging students with enrollment, the student entries get duplicated since it is a one to many relationship where students is the one and enrollment is the many. After merging students with enrollment, the resulting dataframe is merged with coourses in a one to many relationship where courses is the one and the student-enrollment dataframe is the many. Because courses is the one, the entries in courses get duplicated in order to merge with the many in student-enrollment.

```python
na_rows = students_enrollment_courses.loc[students_enrollment_courses.isna().any(axis=1)]

na_rows
```

```python
course_gpas = students_enrollment_courses.pivot_table(
    index='course_id',
    columns='level',
    values='gpa',
    aggfunc='mean'
)

course_gpas = course_gpas.merge(courses_df[['course_id', 'course_title', 'course_number']], 
                                        how='left', on='course_id', validate='one_to_one')

display(course_gpas)

path = os.path.join("..", 'data', 'processed', 'avg_course_gpa.csv')
course_gpas.to_csv(path)
```

I decided to reshape the data by pivotting. I used pivot_table to find the mean gpa for every course, by level, in course_id. I then merged the resulting wide datframe with the courses dataframe to add back important information like the course title and course number.

Replacing every NA value with zeros is a bad idea for a few reasons. If you replace every NA value with zero then when you want to do some kind of computation, like finding the mean, the zero values will influence whatever value you are deriving. Another reason, is that a zero may not be representative of the data. For example, after my pivot I was left with NA mean gpa values. If I replaced these mean gpa values with zeros then somebody reading the table may be confused and think the average gpa for the course is zero when really there is no mean gpa because the course is not offered at that level. Filling all NAs with zeros might also be bad if you have columns with conflicting data types, like a str column.
